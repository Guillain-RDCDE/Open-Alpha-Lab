## 1 · The claims, steelmanned

- **H₁:** YoY gold returns are positively correlated with YoY inflation.
- **H₂:** gold earns more in high-inflation months.
- **H₃:** gold is a safe haven — it rises (not just holds) when equities crash.

## 2 · So what? — what rides on each

H₁/H₂ are the inflation-hedge case; H₃ is the safe-haven case. Failing them doesn't make gold useless — a low/zero correlation is itself valuable for diversification — but it reframes *why* you'd hold it.

## 3 · How we'd know — the protocol

Inflation: corr(YoY gold, YoY CPI) and a high/low (>4%) inflation split. Crisis: corr(gold, stocks) and gold's mean return + up-share in equity-crash months (SPY < −8%).

## 4 · The teardown

### 4.1 The inflation hedge — H₁/H₂

```python
print({k:(round(v,3) if isinstance(v,float) else v) for k,v in ih.items()})
```

> 💡 *In plain words:* corr +0.09, and high-inflation gold (+10.6%) ≈ low-inflation gold (+10.2%). **H₁ and H₂ rejected** — no month-to-month inflation hedge (Erb-Harvey 2013).

### 4.2 The safe haven — H₃

```python
print({k:(round(v,3) if isinstance(v,float) else v) for k,v in cb.items()})
```

> 💡 *In plain words:* gold is near-uncorrelated with stocks (+0.08) and *flat* in the 10 worst equity months (−0.2% vs −10.1%) — but it rose in only **50%** of them. **H₃ partly rejected:** gold is a *hedge* (uncorrelated ballast) but not a reliable *safe haven* (Baur-Lucey 2010's distinction).

## 5 · The verdict

H₁/H₂ rejected, H₃ partly rejected → Signal `NONE` (inflation), Tradability `FRAGILE` (crisis), hedges-inflation `NOT SUPPORTED`. The fingerprinted run is in [`docs/results.md`](../docs/results.md).

## 6 · Could you trade it?

Hold gold for *diversification* — the +0.08 stock correlation and crash-flat behaviour are genuinely useful in a portfolio (it earns its place in risk parity, [68](../../68-all-weather/)). But size it as a diversifier, not as inflation insurance or guaranteed crisis alpha.
