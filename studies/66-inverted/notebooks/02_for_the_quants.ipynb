## Verdict, up front

| Axis | Stamp | Why |

> 💡 *In plain words:* a real warning, a terrible sell button.

## 1 · The claim, steelmanned

- **H₁:** equity returns are lower after an inverted curve.
- **H₂:** the gap is economically large at the recession-lead horizon (18–24m).
- **H₃:** it's a usable *timer* (near-term, reliable).

## 2 · So what? — what rides on each

H₁/H₂ make it a real risk signal; H₃ decides whether you can trade it mechanically. A real-but-slow signal argues for posture, not timing.

## 3 · How we'd know — the protocol

Curve slope → forward S&P return after inverted vs normal at 12/18/24 months → the inverted share and the lead structure.

## 4 · The teardown

### 4.1 Conditional forward returns by horizon

```python
import pandas as pd
display(pd.DataFrame({f'{h}m':st.conditional_forward(slope,d['eq'],h) for h in (12,18,24)}).T[['inverted_fwd','normal_fwd','gap','n_inverted']].round(3))
```

> 💡 *In plain words:* the gap widens from −4% (12m) to −15% (18m) to −30% (24m). **H₁, H₂ hold** strongly.

### 4.2 The variable lead — why H₃ fails

The 12-month gap is only −4.2%: in the year after an inversion, markets are usually still *up*. The damage comes at 18–24 months, but the exact timing varies recession to recession (6–24m). A mechanical 'sell on inversion' rule therefore sits out a year of gains and may be wrong-footed by the melt-up (1998, 2019, 2023). **H₃ rejected** — it's not a timer.

## 5 · The verdict

H₁/H₂ hold, H₃ rejected → Signal `REAL`, Tradability `FRAGILE`, the forecast `CONFIRMED`.

## 6 · Could you trade it?

As a risk-posture input (raise cash / trim beta when the odds rise), not a dated switch. Combine with a confirming trigger (e.g. price below its 200-day) to avoid the year-early exit.
