## 1 · The claim, steelmanned

- **H₁:** risk parity improves risk-adjusted return (Sharpe) vs naïve mixes.
- **H₂:** it cuts drawdown materially.
- **H₃:** it *beats* a stock/bond mix and equities on return (the 'all-weather outperformance' claim).

## 2 · So what? — what rides on each

H₁/H₂ make it a sensible core allocation. H₃ is the marketing promise. Distinguishing risk-adjusted from absolute outperformance is the entire point — they lead to very different portfolios (and very different leverage).

## 3 · How we'd know — the protocol

Inverse-vol weights (60-day vol), monthly rebalance, unlevered → compare ann return / vol / Sharpe / max drawdown against equal-weight, 60/40 and SPY over 2006–2026.

## 4 · The teardown

### 4.1 The four allocations

```python
show=tbl.copy()
for c in ['ann','vol','max_drawdown']: show[c]=(show[c]*100).round(2)
show['sharpe']=show['sharpe'].round(2)
display(show[['ann','vol','sharpe','max_drawdown']])
```

> 💡 *In plain words:* risk parity tops Sharpe (0.92) and drawdown (−17%) and bottoms return (+6.7%). **H₁ and H₂ hold; H₃ fails on absolute return.**

### 4.2 Why H₃ fails — the leverage and bond-bull dependence

Unlevered risk parity caps volatility (~7%), which caps return (~7%). The classic result (Asness-Frazzini-Pedersen 2012) is that risk parity *needs leverage* to be competitive on return — lever to an equity-like vol and the Sharpe edge becomes a return edge. But that imports: (a) financing cost, (b) drawdown risk back to equity levels, (c) a bet that bonds stay low-vol and negatively correlated. 2006–2021 was a historic bond bull that flattered the back-test; 2022 hit stocks *and* bonds together. **H₃ rejected** as an unlevered, regime-free claim.

## 5 · The verdict

H₁/H₂ hold, H₃ rejected → Signal `REAL`, Tradability `INVESTABLE`, beats-everything `OVERSTATED`. The fingerprinted run is in [`docs/results.md`](../docs/results.md).

## 6 · Could you trade it?

Yes, as a calmer core allocation — its drawdown column is the selling point. To chase equity returns you lever it to a vol target, and you must underwrite the financing and the stock-bond correlation regime. Buy it for the ride, size the leverage with eyes open.
