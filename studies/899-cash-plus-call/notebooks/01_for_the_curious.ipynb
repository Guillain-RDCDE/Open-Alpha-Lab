# Study 899 — Cash + Call "90/10" 📈

**Bill Gross's "90/10": park ~90% in T-bills so your capital comes back to par, and spend ~10%
"renting" convex upside with call options. If the market crashes the calls expire worthless and the
bills carry you; if it rockets you get leveraged, capped-loss upside. Does "protect capital, rent
upside" actually beat just holding stocks on a risk-adjusted basis?**

Free listed-option history doesn't exist, so the ~10% convex sleeve is a **documented proxy**: a
rolling **1-year at-the-money SPY call, marked daily with Black–Scholes** (strike = spot at each
annual roll, priced off SPY's trailing realized vol and the ^IRX bill rate; the 10% budget buys as
much notional as the fair price affords, capped near 100% of capital). Real tape SPY/BIL,
2007-05-31 → 2026-06-30.

## 1. Capital protection really works — that part is true

The profile does exactly what it says on the tin: it floors the drawdown. Over 2007–2026 the 90/10 book's worst peak-to-trough was **a third** of buy-and-hold's, at **half the volatility** — the calls simply expired worthless in the crashes while the bills carried the book.

```python
print(f"90/10  : maxDD {R['tt_dd']:.1f}%   vol {R['tt_vol']:.1f}%")
print(f"S&P 500: maxDD {R['bh_dd']:.1f}%   vol {R['bh_vol']:.1f}%")
print(f"2008 crash: S&P {R['crash08_bh']:+.1f}% (DD {R['crash08_bh_dd']:.1f}%)"
      f"  ->  90/10 {R['crash08_c']:+.1f}% (DD {R['crash08_c_dd']:.1f}%)")
```

## 2. …but it doesn't actually *beat* stocks risk-adjusted — it ties

The pitch is a better *risk-adjusted* outcome. It isn't there. Once you measure both books in **excess of cash**, 90/10's **Sharpe 0.49** essentially **ties** buy-and-hold's **0.54** — the gap is a rounding error, and a bootstrap can't tell them apart (P(90/10 wins) ~34%). You gave up **4.6%/yr of compounding** (6.1% vs 10.7%) to reshape the *tail*, not to earn more per unit of risk.

```python
print(f"excess Sharpe : 90/10 {R['tt_sharpe']:.3f}   vs   S&P 500 {R['bh_sharpe']:.3f}   (gap {R['sharpe_vs_bh']:+.3f})")
print(f"CAGR         : 90/10 {R['tt_cagr']:.2f}%   vs   S&P 500 {R['bh_cagr']:.2f}%")
print(f"bootstrap 95% CI on the Sharpe gap: [{R['boot_lo']:+.3f}, {R['boot_hi']:+.3f}]  "
      f"P(90/10 wins) {R['boot_win']:.1f}%  -> a statistical tie")
```

## 3. And a *real* option costs more than our model assumed

Our call is priced off **realized** volatility. A real listed call trades at **implied** vol, which is systematically *higher* (the variance risk premium — that's how option sellers get paid). Charge the option what it actually costs — a **1.25–1.5×** markup — and the tie flips to a clear **loss** vs buy-and-hold. Renting upside is a negative-carry trade priced to leave the buyer behind.

```python
for lbl,sh,vs in [('BS-fair (realized vol)',R['pm100_sh'],R['pm100_vs']),
                  ('1.25x (mild VRP)',R['pm125_sh'],R['pm125_vs']),
                  ('1.50x (typical)',R['pm150_sh'],R['pm150_vs'])]:
    print(f"  option cost {lbl:24s}: 90/10 Sharpe {sh:+.3f}   vs S&P {vs:+.3f}")
```

## 5. The honest verdict

- **Signal: Weak.** The capital protection is **real and mechanical** (maxDD -19.8% vs -55.2%, 2008 cut from −47% to −10%). But the *risk-adjusted-return* claim isn't there: excess Sharpe **0.49 vs 0.54** is a statistical tie, the option's convexity adds no alpha (*t* = +0.33), and that's at a premium that *flatters* the strategy.
- **Tradability: Mirage.** The parity survives only at the fictional fair price — a realistic option markup (1.25–1.5×) sinks it clearly below buy-and-hold. It's not a trading-cost story (one roll a year); it's that renting upside is negative-carry. You run 90/10 to **sleep at night**, not to get paid.
