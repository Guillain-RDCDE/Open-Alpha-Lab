## Verdict

- **Signal — None.** The industry-relative MAX does **not** sharpen a valid MAX effect on 50 liquid US mega-caps — because the effect is *inverted* here. The Q1−Q5 spread is **-89.7 bps/mo** (NW *t* = **-2.51**), significant but *opposite in sign* to the claim (the sign-flip placebo puts it ~2.8σ into the left tail), and it holds only in the late era (*t* = -1.26 early / -2.24 late). A significant wrong-sign result **fails the claim**. The 20-seed synthetic control recovers a *planted* relation cleanly and confirms the adjustment sharpens it (raw *t* = +11.18 → ind-rel *t* = +20.60), so the sign-reversal is real, not machinery. Survivorship biases the magnitude.
- **Tradability — Mirage.** The specified long-low / short-high book loses money gross (**-89.7 bps/mo**) and net (**-95.9** at 1 bp, **-103.9** at 5 bps). A Mirage in the claimed direction.
