```python
import sys, os
sys.path.insert(0, os.path.abspath("../../.."))  # repo root (quantlab/)
sys.path.insert(0, os.path.abspath(".."))        # study package (bunker/)
%matplotlib inline
import matplotlib.pyplot as plt
plt.rcParams["figure.figsize"] = (10, 5.5); plt.rcParams["axes.grid"] = True
import numpy as np, pandas as pd
from bunker import data, strategy as st
ret = data.fetch_pairs()                       # cache-first
al = ret[["USMV","SPY"]].dropna()              # common window (USMV starts 2011)
sp = st.spread(ret, "USMV", "SPY")
```

## Verdict, up front

| Axis | Stamp | Why |
|---|---|---|
| Signal | **Weak** | vol 0.80×, DD −20% vs −24%; Sharpe 0.99 < 1.05 |
| Tradability | **Fragile** | trailed return (−3.5%/yr); defensive only |
| Sharpe edge? | **Not supported** | absent in a bull sample |

> 💡 *In plain words:* it delivers calm, not a free Sharpe.

## 1 · The claim, steelmanned

- **H₁:** USMV cuts volatility/drawdown vs the market.
- **H₂:** USMV beats the market on Sharpe.
- **H₃:** USMV beats it on return.

## 2 · So what? — what rides on each

H₁ is the defensive case; H₂ is the anomaly's promise. If only H₁ holds, USMV is risk management, not alpha.

## 3 · How we'd know — the protocol

Leg CAGR/Sharpe/vol/drawdown over the common window → vol-reduction ratio → the spread and its t-stat.

## 4 · The teardown

### 4.1 The two legs

```python
import pandas as pd
display(pd.DataFrame({'USMV':st.leg_summary(al,'USMV'),'SPY':st.leg_summary(al,'SPY')}).T[['cagr','sharpe','vol_ann','max_drawdown']].round(3))
print('vol reduction:', round(st.vol_reduction(al,'USMV','SPY'),2))
```

> 💡 *In plain words:* **H₁ holds** (0.80× vol, shallower DD) but **H₂ and H₃ rejected** (Sharpe 0.99 < 1.05, less return).

### 4.2 The spread

```python
print({k:round(v,3) for k,v in st.spread_stats(sp).items()})
```

> 💡 *In plain words:* −3.5%/yr — USMV trailed; the low-vol Sharpe edge is absent here.

### 4.3 The regime caveat

2011–2026 was largely a strong-market, high-beta regime. The low-vol anomaly historically earns its keep in **bear markets and over full cycles** (it loses less when stocks fall). A sample without a sustained bear under-states it — the same regime that made high-beta *win* in [43 Free-Lunch](../../43-free-lunch/) and [53 Jackpot](../../53-jackpot/). Stated, not hidden.

## 5 · The verdict

H₁ holds, H₂/H₃ rejected → Signal `WEAK`, Tradability `FRAGILE`, the Sharpe edge `NOT SUPPORTED` (this sample).

## 6 · Could you trade it?

As a defensive sleeve, yes — lower volatility and drawdown are real and useful. As a market-beater, no, at least not in a bull-dominated window.
