## The accounting, in one line

`after_tax = price x (1 - capgain) + income x (1 - tau_income)`, with the income leg reconstructed as `total_return - price_return` per month. `tau_income` is 0 for munis federally (state only, and only on the out-of-state slice), `fed + NIIT + state` for taxable credit, `fed + NIIT` for T-bills. With the default `capgain = 0` this collapses to `total - tau x income`, so the after-tax difference between a muni and a taxable leg is **exactly affine in tau**:

`d(tau) = d(0) + tau x mean(income_taxable)`

which is why the break-even needs no search: `tau* = -mean(d(0)) / mean(income_taxable)`.

> 💡 **In plain words.** Raise the tax rate and the only thing that changes is how much the taxable bond's coupon is docked. That makes the muni-minus-taxable gap a straight line, and the break-even is just where the line crosses zero.

## Break-even by pairing — with the interval that decides whether to quote it

`tau*` is a **ratio of two sample means**, and its numerator (the pre-tax total-return difference) never clears |*t*| ≥ 2 in any pairing on this tape. The point estimate is therefore far more precise-looking than the data warrants. The block bootstrap below is the correction; `p>40.8%` is the share of resamples in which **no US bracket is high enough**.

```python
rows = [('MUB','VCIT', R['pre_mub_vcit'], R['pre_mub_vcit_t'], R['be_mub_vcit'], R['beci_mub_vcit']),
        ('MUB','LQD ', R['pre_mub_lqd'],  R['pre_mub_lqd_t'],  R['be_mub_lqd'],  R['beci_mub_lqd']),
        ('VTEB','VCIT', None, None, R['be_vteb_vcit'], R['beci_vteb_vcit']),
        ('MUB','AGG ', R['pre_mub_agg'],  R['pre_mub_agg_t'],  R['be_mub_agg'],  R['beci_mub_agg']),
        ('VTEB','AGG ', None, None, R['be_vteb_agg'], R['beci_vteb_agg']),
        ('SUB','BIL ', None, None, R['be_sub_bil'], R['beci_sub_bil']),
        ('HYD','LQD ', None, None, R['be_hyd_lqd'], R['beci_hyd_lqd'])]
for a, b, pre, t, be, ci in rows:
    pre_s = '   n/a        ' if pre is None else 'pre-tax %+6.2f bps (t=%+5.2f)' % (pre, t)
    print('%-5s vs %-5s  %s  tau* %+6.1f%%  95%% CI [%+7.1f%%, %+6.1f%%]'
          % (a, b, pre_s, be, ci[0], ci[1]))
print()
print('MUB/VCIT: %.0f%% of resamples put tau* ABOVE the top US rate (40.8%%),'
      % (100*R['p_above_top_mub_vcit']))
print('          %.0f%% put it BELOW zero. The total-return break-even is not identified.'
      % (100*R['p_below_zero_mub_vcit']))
print('The width is the PRICE legs: MUB-VCIT price diff %+.2f bps/mo (t=%+.2f).'
      % (R['price_leg_mub_vcit'], R['price_leg_mub_vcit_t']))
```

> ⚠️ **The sweep fallacy.** `tau*` is invariant to the bracket, the state rate, the in-state share and the capital-gains rate — because *no tax parameter enters its numerator*. "Stable under every sweep" is a statement about the arithmetic and carries **zero** information about precision. Only resampling the tape does (`strategy.breakeven_ci`, regression-tested both ways).

## Where the *t*-stat comes from — `d(tau) = d(0) + tau x i_taxable`

The second term is a coupon stream: large in the mean, near-constant in time. Adding a near-deterministic constant to a noisy series lifts the mean without lifting the variance, so **HAC *t* climbs with the bracket by construction**.

```python
print('pair        pre-tax (t)      + tax term (sd)     = total (t)      mean%   var%')
for name, d in [('MUB-AGG ', R['dec_mub_agg']), ('VTEB-AGG', R['dec_vteb_agg']),
                ('SUB-BIL ', R['dec_sub_bil']), ('HYD-LQD ', R['dec_hyd_lqd']),
                ('MUB-VCIT', R['dec_mub_vcit'])]:
    print('%s  %+6.2f (%+5.2f)   %+6.2f (sd %4.2f)   %+6.2f (%+5.2f)  %6s  %5.2f'
          % (name, d[0], d[1], d[2], d[3], d[4], d[5],
             '--' if d[6] is None else '%.1f' % d[6], d[7]))
print()
print('Every row crossing |t| = 2 does so on a term carrying < 2% of the variance.')
print('That is a test of whether the tax code is nonzero, not of a market.')
```

## Costs (irrelevant) and borrow (fatal)

```python
print('cost  0 bps one-way: %+5.2f bps/mo' % R['cost0'])
print('cost 25 bps one-way: %+5.2f bps/mo   <- one trade in 17 years' % R['cost25'])
print()
print('borrow   0 bps/yr: MUB-VCIT %+6.2f   MUB-AGG %+6.2f' % (R['borrow0_vcit'], R['borrow0_agg']))
print('borrow  25 bps/yr: MUB-VCIT %+6.2f   MUB-AGG %+6.2f (t=%+.2f)  <- both broken'
      % (R['borrow25_vcit'], R['borrow25_agg'], R['borrow25_agg_t']))
print('borrow 100 bps/yr: MUB-VCIT %+6.2f   MUB-AGG %+6.2f' % (R['borrow100_vcit'], R['borrow100_agg']))
```

> 💡 **In plain words.** Choosing which bond fund to own costs one trade, so friction is a rounding error. Trying to *arbitrage* the pair means shorting the taxable fund, and the borrow fee on a corporate-bond ETF is larger than the entire edge.

## Live synthetic control — machinery proof, never a stamp

Forty years of monthly data, eight seeds. **Planted world:** the taxable leg yields 150 bp more pre-tax, so the true break-even is exactly 33.33% and the after-tax difference must flip sign across it. **Twin null:** identical yields and duration, so the break-even must collapse to ~0 and no *pre-tax* difference may be manufactured.

Note what the null deliberately does *not* claim: on twins the *after-tax* difference at a positive bracket is large and significant, because it is pure arithmetic (`tau × yield`). That is the effect being measured; the control's job is to show the machinery attributes none of it to a market edge.

```python
import os, sys
sys.path.insert(0, os.path.abspath('..'))
sys.path.insert(0, os.path.abspath(os.path.join('..','..','..')))
import numpy as np
from after_tax import data, strategy as st
for ss, label in [(1.0, 'planted 150bp gap'), (0.0, 'twin null      ')]:
    errs = []
    for s in range(8):
        panel, truth = data.synthetic_panel(signal_strength=ss, seed=952 + s)
        errs.append(st.breakeven_rate(panel, 'muni', 'taxable')['breakeven']
                    - truth['planted_breakeven'])
    panel, truth = data.synthetic_panel(signal_strength=ss, seed=952)
    d = st.synthetic_detect(panel)
    errs = np.array(errs)
    print('%s true %6.2f%%  recovered %6.2f%%  (8 seeds: mean err %+.2f pp, max |err| %.2f pp)'
          % (label, truth['planted_breakeven']*100, d['breakeven']*100,
             errs.mean()*100, np.abs(errs).max()*100))
    print('    pre-tax diff %+6.2f bps (t=%+5.2f)   below break-even %+6.2f bps -> above %+6.2f bps'
          % (d['pretax_diff_bps'], d['pretax_t'],
             d['diff_below_breakeven'], d['diff_above_breakeven']))
```

### The t-stat demonstration, live on the twin null

Two **statistically identical** bonds, nothing planted. Watch the after-tax HAC *t* walk past 2 as the assumed bracket rises. Nothing arrives but arithmetic — which is why no |*t*| ≥ 2 in the real-tape tables above can be read as a market effect.

```python
panel, _ = data.synthetic_panel(signal_strength=0.0, seed=952)
print('twin null: identical yields, identical duration, zero planted edge')
for f in (0.0, 0.10, 0.20, 0.30, 0.40):
    r = st.race(panel, 'muni', 'taxable', st.tax_profile(f, 0.0, 0.0),
                cash='cash', cost_bps=0.0)
    print('  bracket %4.0f%%   after-tax diff %+7.2f bps/mo   HAC t = %+5.2f'
          % (100*f, r['diff_bps'], r['t_diff']))
dec = st.tax_constant_decomposition(panel, 'muni', 'taxable', st.tax_profile(0.37))
print('\nat 40.8%%: tax term is %.1f%% of the mean and %.2f%% of the variance'
      % (100*dec['mean_share_tax'], 100*dec['var_share_tax']))
```

## Verdict

- **Signal — Weak.** Exactly one estimand here is identified, and it is not an outperformance. The **income-leg break-even** is tight — **26.7%** [23.4, 29.7] vs VCIT, **28.9%** [27.1, 30.7] vs LQD — measured on realised distributions net of fees, and it sits below the top two brackets. Everything built on top of it fails. The **total-return break-even is not identified**: 35.0% point, 95% CI [-10.9%, 81.5%], 41% of draws above the top US rate — its numerator never clears |*t*| ≥ 2 in any pairing (largest -1.50). The **after-tax race is a dead heat**: +1.72 bps/mo (+0.21 pp/yr), HAC *t* = +0.25, CI [-12.5, +15.6]. And the three pairings clearing |*t*| ≥ 2 clear it on a **tax constant** worth 86% of the mean and 0.30% of the variance — the same lift appears on the synthetic twin null with nothing planted. **No |*t*| ≥ 2 in this study is evidence of a market effect**, and none survives either era half. The control recovers a planted 33.33% break-even to within 10.3 pp across 8 seeds and stays quiet on twins (pre-tax *t* = -1.46), so the dead heat is the muni market, not the harness.
- **Tradability — Fragile.** Real and free as an *allocation* (one round trip, 0.03 bps/mo amortised; in a top bracket hold munis in the taxable account, which the tight income-leg break-even does support), but the total-return margin sits inside the noise, the answer rides on an assumed bracket, the crossover cannot be pinned better than [-11%, 82%], and as a *trade* -0.37 bps/mo at 25 bps/yr borrow is the whole story.
