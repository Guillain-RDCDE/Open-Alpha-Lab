# Study 915 — K-1 vs 1099 🧾

**Does the tax-friendly commodity wrapper give back its convenience in performance?**

Buy a commodity ETP that holds futures directly and you own a **partnership**. Every
spring it sends you a **Schedule K-1** — often late enough to force a filing extension —
and inside an IRA it can generate taxable income the account was supposed to shelter. In
exchange, its futures gains fall under **Section 1256**: taxed **60% long-term / 40%
short-term** whatever your holding period, but **marked to market every 31 December**
whether or not you sell.

**PDBC** is the same manager's answer, branded literally *"No K-1"*. It pushes the futures
into a Cayman subsidiary so it can be a normal 1940-Act fund and send you a **Form 1099**.
The trade: you lose the 60/40 rate, you receive ordinary-income distributions — and you
gain **deferral** on everything else.

The desk's question is not which structure is prettier. It is whether the convenience shows
up as a **cost on the tape**. We race **PDBC (1099)** against **DBC (K-1)** — same manager,
same index family, same commodity beta — on daily **total-return** closes,
2014-11-07 → 2026-06-30 (2,926 days), both legs **excess of cash** (BIL).

*Numbers below are the frozen headline (`docs/results.md`, Fingerprint `0e33b87210d5`); the
only live cells run a synthetic control and say so. As-of 2026-06-30.*

## 2. The answer: nothing to see

The 1099 wrapper trailed its K-1 twin by **0.13% a year** — about a tenth of one percent. That is not a finding; it is a rounding error. The *t*-statistic is **-0.21** (anything under 2 means "indistinguishable from zero"), and the honest range around it runs from **-1.03% to +0.78%** a year — comfortably straddling nothing at all.

On risk-adjusted terms the gap is even smaller: an excess-of-cash Sharpe difference of **-0.0081**, with a range of [-0.059, +0.042]. For practical purposes these are the same investment wearing different paperwork.

## 3. "But my null might just be blind"

A finding of *nothing* is only worth reading if the test could have found *something*. So we asked the sample how big a wrapper cost it would have caught. Answer: anything above roughly **0.9-1.2% a year** would have shown up clearly.

So the correct sentence is not "there is no cost" — it is **"there is no cost bigger than about 1% a year"**. Given that the entire fee difference between the two funds is 28 basis points, that is a useful bound.

> 🔬 **For the quants:** the minimum detectable difference at |*t*| = 2 is 1.22%/yr on the HAC standard error and 0.91%/yr on the 21-day block bootstrap; the naive i.i.d. figure (3.11%/yr) is three times too pessimistic because the daily difference is strongly *negatively* autocorrelated.

## 4. The 5% "tracking error" that isn't

Measured day by day, the two funds look like they wander 5.3% apart a year — which sounds alarming. Measure the same thing month by month and it collapses to **1.46%**.

The reason is mundane: the two closing prints do not clear at the same instant, and each carries its own small premium or discount to the fund's true value. They disagree on Monday and re-converge on Tuesday — the daily difference has a lag-1 correlation of **-0.43**. That is stock-exchange plumbing, not divergence a holder ever experiences.

## 5. Year by year — and the one year that looks like a story

Across the 11 complete calendar years the 1099 wrapper won **3 of 11** (27%, and a coin flip sits inside the honest range of 10-57%). Nine of the eleven annual differences are inside ±1 percentage point.

The exception is **2025**, where PDBC trailed by **2.14 pp**. PDBC is *actively* managed — it does not mechanically copy DBC's index — so it is allowed to hold a different basket, and manager drift is the natural explanation. Be clear that this is an **interpretation, not a measurement**: we did not open the holdings files, so we cannot prove it was the basket rather than the envelope. What we can say is that the gap points the other way in 2015 (+0.74 pp), which is not how a structural wrapper cost behaves.

## 6. Now the tax — and the honest bit

Pre-tax, the two are a tie. So does tax break the tie? We modelled both regimes on the realised annual returns: the K-1 leg marked to market every year at the 60/40 rate, the 1099 leg paying ordinary tax on its distributions and deferring the rest until sale.

**Every input to that model is an assumption, not a measurement** — the marginal rates, and above all how much the 1099 fund distributes each year. So we swept both. The result across the whole grid runs from **-0.36 to +0.59 percentage points a year** — and **the sign flips inside it**.

If the fund distributes little, deferral wins and the 1099 wrapper is ahead by a few tenths of a point. If it distributes twice the collateral interest — which a fund of this type can do when its subsidiary realises gains, and which we did **not** measure for PDBC — the K-1's 60/40 rate wins by about the same margin. The after-tax winner is chosen by *your assumption about the distribution*, not by the tape. We report the range and refuse to pick.

## 7. The decision you are actually making

Over the same window, choosing between commodity vehicles was worth **11.4 percentage points a year**: USCI compounded at +4.68%, USO at **-6.70%**. Choosing between the two *wrappers* was worth **0.13 points**.

The paperwork decision is roughly **90 times smaller** than the decision it is usually bundled with — and investors spend far more time on it.

## Verdict

- **Signal — None.** The 1099 wrapper costs **0.13%/yr** against its K-1 twin, with *t* = -0.21 and a range of [-1.03%, +0.78%]; null in both halves of the sample, 3/11 annual wins, and unmoved by a punitive cost sweep. The test had the power to catch anything above ~1%/yr, so this is an informative null. The after-tax model does not rescue a signal either: its gap spans [-0.36, +0.59] pp/yr and **changes sign** with the distribution assumption.
- **Tradability — Mirage.** There is no edge here to bank — the difference is tens of basis points against 1.46-5.30% of tracking error and 11.4 pp/yr of index-choice dispersion. What *is* worth knowing is the non-finding: the No-K-1 convenience appears to be **free**. That is a good reason to prefer it, and it is not alpha.
