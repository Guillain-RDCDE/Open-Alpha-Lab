## Verdict, up front

| Axis | Stamp | Why |
|---|---|---|
| Signal | **None** | VYM Sharpe 0.67 < SPY 0.76; spread −1.5%/yr |

| Tilt beats? | **Busted** | payout form ≠ premium (Miller-Modigliani) |

> 💡 *In plain words:* a sector bet wearing an income halo.

## 1 · The claim, steelmanned

- **H₁:** high-dividend total return beats the market.
- **H₂:** it does so risk-adjusted (higher Sharpe / lower drawdown).
- **H₃:** the spread is positive.

## 2 · So what? — what rides on each

If any hold, the income tilt is justified. If none do, it's a costly behavioural preference for the *form* of returns.

## 3 · How we'd know — the protocol

Leg CAGR/Sharpe/drawdown → the spread and its Lo t-stat.

## 4 · The teardown

### 4.1 The two legs

```python
import pandas as pd
display(pd.DataFrame({'VYM':st.leg_summary(ret,'VYM'),'SPY':st.leg_summary(ret,'SPY')}).T[['cagr','sharpe','vol_ann','max_drawdown']].round(3))
```

> 💡 *In plain words:* VYM trails on CAGR *and* Sharpe, with a deeper drawdown. **H₁, H₂ rejected.**

### 4.2 The spread

```python
print({k:round(v,3) for k,v in st.spread_stats(sp).items()})
```

> 💡 *In plain words:* −1.5%/yr, negative more than half the months. **H₃ rejected.**

### 4.3 Why — dividends are payout form, not a premium

Miller-Modigliani (1961): payout policy doesn't create value. A high yield changes the *form* of return (cash vs capital gain), not the amount — and a total-return investor is indifferent. The high-yield *screen*, meanwhile, loads on mature value sectors (banks, utilities, energy) that lagged the tech-led 2007–2026 market. A sector bet, not an edge.

## 5 · The verdict

H₁, H₂, H₃ all rejected → Signal `NONE`, Tradability `MIRAGE`, the tilt `BUSTED`.

## 6 · Could you trade it?

No edge. For index income, a 'home-made dividend' (selling shares) gives the same cash without the value/sector tilt or the yield-trap risk.
