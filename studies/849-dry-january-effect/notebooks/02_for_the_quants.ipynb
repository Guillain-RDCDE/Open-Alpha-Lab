## Verdict

- **Signal — Weak.** The Dry-January / Veganuary footprint is *directionally perfect* — alcohol is the year's most-negative calendar month (-1.20%, rank 1/12) and `BYND` the most-positive (rank 1/12), sign-consistent across both eras — but the only well-sampled leg reaches just *t* = -1.28 (27 Januaries), the plant/spread side rides 7 noisy `BYND` years (*t* = +1.33), the February hangover is absent, and **no cut clears |*t*| ≥ 2**. The 20-seed control fires 90% on a genuine +5% seasonal and 0/20 on the null, so the whiff is real-but-sub-threshold.
- **Tradability — Mirage.** The long-plant / short-alcohol January timer nets **+14.22%/yr** at 5 bps — but on 7 observations at *t* = +1.31, driven by one small-cap's early tape, capacity-trivial and inside a multi-cut search. Nothing bankable.
