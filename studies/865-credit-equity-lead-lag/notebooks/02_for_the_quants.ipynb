## Verdict

- **Signal — None.** The credit trend does not lead next-week equity: the predictive-regression slope is wrong-signed and insignificant at every horizon (**-19.6 bps/σ**, NW *t* = **-1.70** at 4 weeks; *t* = -1.58 at 1 week; *t* = -0.23 at 2 weeks), R² never exceeds 0.6%, the risk-on−risk-off difference sits -0.4σ inside a label-shift placebo (p = 0.66), and the wrong sign is consistent across eras. The 20-seed synthetic control recovers a *planted* lead (*t* = +26.7, fires on 0/20 nulls), so the flat real-tape result is a true absence.
- **Tradability — Mirage.** The costed SPY↔IEF overlay never beats buy-and-hold (net Sharpe 0.644 vs 0.645 at 1 bp, active NW *t* = -1.22; it forfeits 3.3%/yr of CAGR) and falls further behind by 5 bps. Its one merit — a -33% vs -55% drawdown — is insurance paid for in forgone return, not an edge.
