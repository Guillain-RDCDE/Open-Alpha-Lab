> 💡 *In plain words:* a genuine quality screen, measured in the wrong universe.

## 1 · The claim, steelmanned

- **H₁:** high F-score predicts high forward returns.
- **H₂:** the long-high / short-low hedge is positive and significant.
- **H₃:** it's tradable on liquid large caps (not just small-cap value).

## 2 · So what? — what rides on each

H₁/H₂ would make the F-score a free, transparent quality factor straight from 10-Ks. H₃ decides whether it survives where you can trade in size. A small-cap-value-only effect is real science but useless to a large-cap book.

## 3 · How we'd know — the protocol

Nine EDGAR concepts → the 0–9 F-score → annual tercile sort → long top (healthy), short bottom (weak) → next-year hedge, Sharpe, t-stat, hit rate over the XBRL window.

## 4 · The teardown

### 4.1 The long-short, year by year

```python
disp=h.copy(); disp.columns=['high-F leg','low-F leg','hedge (high-low)']
display((disp*100).round(1))
s=st.summary(h['hedge']); print({k:round(v,3) for k,v in s.items()})
```

> 💡 *In plain words:* the hedge mean is **negative** (−3.7%/yr) and wins only ~33% of years. **H₁ and H₂ are rejected on this universe** — the sign is backwards.

### 4.2 Why it doesn't replicate — H₃ fails

Four forces flatten and flip it on large caps:

1. **It's a small-cap value effect.** Piotroski (2000) built the F-score *within* high book-to-market firms — small, neglected, distressed. That's where the spread lives.
2. **High-F is crowded.** On heavily-covered large caps a strong balance sheet is already in the price — no neglect premium left.
3. **Low-F snaps back.** Beaten-down large-cap cyclicals / troughing growth rebound, so the *short* leg wins.
4. **Survivorship.** Current members exclude the low-F firms that actually went bankrupt — the ones the short was meant to catch. **H₃ rejected.**

## 5 · The verdict

H₁/H₂/H₃ all rejected on tradable large caps → Signal `WEAK` (inverted), Tradability `MIRAGE`, sorts-large-cap-winners `NOT SUPPORTED`. The fingerprinted run is in [`docs/results.md`](../docs/results.md).

## 6 · Could you trade it?

Not on this universe. To harvest the F-score premium you'd run it on a small-cap, high-book-to-market book — illiquid, under-covered names where a clean scorecard is a genuine surprise. Outside what a free EDGAR pull on current S&P members offers.
