## The race — CPPI vs buy-and-hold vs matched static mix (excess-of-cash, gross)

A *constant* fraction of SPY funded from cash has the **same** excess-of-cash Sharpe as SPY — so the matched-static and buy-and-hold Sharpes coincide, and CPPI's only distinguishing act is the dynamic **re-timing** (the spanning alpha).

```python
print(f"CPPI        : excess-Sharpe {R['cppi_sharpe']:.3f}  CAGR {R['cppi_cagr']:5.2f}%  "
      f"vol {R['cppi_vol']:5.2f}%  maxDD {R['cppi_dd']:.1f}%")
print(f"buy-and-hold: excess-Sharpe {R['bh_sharpe']:.3f}  CAGR {R['bh_cagr']:5.2f}%  "
      f"vol {R['bh_vol']:5.2f}%  maxDD {R['bh_dd']:.1f}%")
print(f"static@avgw : excess-Sharpe {R['static_sharpe']:.3f}  CAGR {R['static_cagr']:5.2f}%  "
      f"vol {R['static_vol']:5.2f}%  maxDD {R['static_dd']:.1f}%")
print(f"vs buy-and-hold {R['sharpe_vs_bh']:+.3f}   re-timing alpha {R['alpha_ann']:+.2f}%/yr "
      f"(HAC t {R['t_alpha']:+.2f}, beta {R['beta']:.3f})   avg-w {R['avg_w']:.2f}, "
      f"turnover {R['turnover']:.1f}x/yr, breaches {R['n_breach']}")
```

## Costed & gap risk — no return edge to erode, and the floor isn't a hard guarantee

CPPI trails buy-and-hold **at zero cost** (not a friction story); and a one-day drop past **1/m = 20%** breaches the floor — daily rebalancing can't pre-empt an overnight gap.

## Verdict

- **Signal — MIXED.** The floor-protection claim is **real and mechanical**: maxDD **-20.8%** vs buy-and-hold's **-55.2%**, **0** breaches, 2008 cut from −47% to −11%, confirmed by a 30-seed synthetic control. But the implied *'insurance improves the risk-adjusted outcome'* claim is **false**: excess Sharpe **0.164 vs 0.542** (bootstrap gain -0.379, 95% CI [-0.691, -0.050]), no multiplier or floor beats buy-and-hold, and the re-timing spanning alpha **subtracts** value (-1.56%/yr, *t* = -2.38 in the volatile era). A single-cycle, GFC-anchored, ~19-year window.
- **Tradability — MIRAGE.** No bankable *return* edge — CPPI trails buy-and-hold gross, before costs. The drawdown cap is cheap and genuine **as insurance** (2.3×/yr turnover on SPY, survives every cost) but it is a *cost* paid in full out of upside (2.4% CAGR vs 10.7%), not a paycheck — and an overnight gap past 1/m = 20% breaches even the floor.
