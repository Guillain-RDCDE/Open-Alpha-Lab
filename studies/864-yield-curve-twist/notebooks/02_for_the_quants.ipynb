## Synthetic positive control — unbiased, but a HAC caveat

Live: the detector must recover a planted edge and (near-)silence on the null. The null *t* has sd > 1 — the HAC test over-rejects under a persistent regressor + overlapping returns, so the real full-sample t is discounted.

```python
import os, sys
sys.path.insert(0, os.path.abspath('..'))
sys.path.insert(0, os.path.abspath(os.path.join('..','..','..')))
import numpy as np
from curve_twist import data, strategy as st
planted = st.synthetic_detect(data.synthetic_daily(n_days=4000, fly_signal=0.02, seed=864)[0], 21)
print(f"planted (fly_signal=0.02): reg t = {planted['t']:+.2f}, Q5-Q1 t = {planted['t_spread']:+.2f}")
null_t = np.array([st.synthetic_detect(data.synthetic_daily(n_days=2500, fly_signal=0.0, seed=864+s)[0], 21)['t'] for s in range(8)])
print(f"null (fly_signal=0), 8 seeds: reg t mean {null_t.mean():+.2f} (sd {null_t.std(ddof=1):.2f}), |t|>=2 in {(abs(null_t)>=2).sum()}/8")
print('  sd > 1 -> HAC over-rejects; the real t=2.34 is ~1.8 effective sigma.')
```

## Verdict

- **Signal — Weak.** The butterfly predicts forward belly-Treasury (IEF) returns with the right sign and a full-sample HAC *t* = **+2.34** (β = **+16.9 bps/1σ**), and it is *not* the 2s10s slope repackaged (slope control *t* = +0.25). But it **fails the era bar**: all the juice is in 2002-2009 (*t* = +3.32), dead after (*t* = -0.18 / +0.59); the quintile *t* is 1.66, the incremental *t* is 1.98, the *twist* is null, and the HAC *t* is inflated (~1.8 effective σ).
- **Tradability — Mirage.** The curvature timer loses to buy-and-hold on mean return at every cost (-0.39 to -0.61 bps/day); the Sharpe edge is a cash-parking volatility artefact.
- **Twist beyond level+slope? — Mixed.** Standing curvature holds marginal (unstable) info the slope does not; its *change* holds none.
