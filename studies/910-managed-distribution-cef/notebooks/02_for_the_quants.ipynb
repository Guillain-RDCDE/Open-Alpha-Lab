Live: the null (pure levered beta) must NOT fire; a planted net carry must be recovered as alpha; a 100%-return-of-capital payout must yield no alpha.

- **Signal — Weak.** The payout is a *real* excess-of-cash return (basket **+88.3 bps/mo**, HAC *t* = **+2.63**, bootstrap Sharpe CI [0.12, 1.17] clear of zero) — not the capital shredder an mREIT is. But the **asset-class edge fails**: the excess-vs-excess Sharpe trails SPY in every fund and sub-era (basket adv **-0.21**, -0.10 pre / -0.36 post; PCEF -0.41), the CAPM alpha is ~0-to-negative (-1.6 %/yr; PCEF -3.3 %/yr, *t* = -1.96), and β ≈ 1.00 exposes a levered-beta income clone. The synthetic control confirms the machinery is unbiased.
- **Tradability — Mirage.** Net ≈ gross (+88.0 bps; buy-and-hold), so costs don't kill it — a **hidden levered equity beta** does: SPY's Sharpe (0.82) beats the basket's (0.61), and the 2022 rate shock turned the excess return to +32 bps (*t* = +0.62) with a -5.5 %/yr alpha and a -24.4 % year.
