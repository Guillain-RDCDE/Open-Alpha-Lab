# Study 929 — Rights Offering 🎟️

**Is the deep subscription discount on a rights issue a gift, or a dilution warning?**

When a listed company or fund runs a **rights offering**, every existing holder is handed
the right to buy new shares at a price well below the market — 20% or more below, in the
small closed-end funds and BDCs that dominate US rights activity. That looks like free
money. It usually isn't: the rights go out *pro rata*, so a holder who subscribes in full
owns the same slice of a slightly larger pie, and the share price mechanically falls by
about the value of the right. The discount is a bookkeeping choice made to guarantee
take-up, not a transfer.

We test both readings on **39 US rights offerings** by **20
issuers** (2013-05 → 2023-09), measuring abnormal returns against
**SPY** around the announcement, across the subscription period and after expiry.

*Real-tape numbers below are the frozen headline (`docs/results.md`, Fingerprint
`9acc9535dfdd`, as-of 2026-06-30); the live cells run the fast **synthetic** control and are
labelled as such.*


## 1. Why a discount you are *given* can't make you richer

Imagine a fund worth £100 a share, and every holder is offered one new share at £80 for each two they own. Afterwards the fund holds more cash per share than before but has more shares outstanding; the price settles near £93. If you subscribed, you paid £80 for something now worth £93 — but your existing two shares each fell from £100 to £93. You are exactly where you started.

The only person the discount actually *costs* is the holder who ignores the letter: they get the price fall without the cheap share. That is the whole economics of a rights issue. The interesting empirical question is what is left over — does the market read the *decision* to raise equity as bad news, and does a deeper discount signal a worse issuer?

> 🔬 **For the quants** — the theoretical ex-rights price is the pre-deal market cap plus the cash raised, divided by the post-deal share count. Yahoo's `auto_adjust` back-adjusts for splits and dividends but **not** for a rights issue, so that mechanical fall stays in the tape as a raw price drop. Any window spanning the ex-rights date therefore mixes news with arithmetic — which is why our verdict is formed on the announcement and post-expiry windows, not the subscription window.

## 4. The one number that looked real

Across the subscription period the deals lose **-2.06%** against SPY with a *t* of **-2.03** — nominally significant. Two reasons not to believe it.

**First**, that window is exactly where the mechanical ex-rights drop lives (section 1), so a fall there is arithmetic, not news.

**Second**, it is not robust. Drop one issuer's five deals and the *t* falls to -1.14; split the sample at 2018 and the recent half gives -1.05; change the assumed expiry date from 40 to 28 days and it is -1.43; shift the (month-precision) anchors by a fortnight and it never clears −2 with room to spare. A result that needs one issuer, one decade and one guessed date to exist is not a finding.

> 🔬 **For the quants — and a caution about our own favourite chart.** Re-run the study on 300 sets of *random* dates and the mean abnormal return over a 28-day window wanders with a standard deviation of 2.03%, which would make our result a *z* of -0.93 — a nothing. But that placebo draws its dates from the whole 2005-2026 tape, including 2008 and 2020, regimes in which none of these deals was announced: it is a rigged-easy yardstick. Restrict the draws to the years the deals actually happened and the dispersion falls to 1.28%, giving *z* = -1.60 (*p* = 0.09) — still not significant, but not the rout the first version implied. We quote the harder number.

## 5. Could you trade it either way?

Buy every announcing name the day *after* the news and hold 49 sessions, 25 bps one-way, everything else in T-bills. Because that holding period straddles the ex-rights date — where the price mechanically falls for reasons that are arithmetic, not news — we also run the book over the clean stretch *after* the deal closes, which is the version that gives the discount its best shot:

```python
print(f"long,  full  window : excess Sharpe {R['long_sharpe']:+.3f}, "
      f"worst loss {R['long_dd']:.1f}%")
print(f"long,  clean window : excess Sharpe {R['cln_long_sharpe']:+.3f}, "
      f"worst loss {R['cln_long_dd']:.1f}%")
print(f"short, full  window : excess Sharpe {R['short_sharpe']:+.3f}")
print(f"just holding SPY    : excess Sharpe {R['spy_sharpe']:+.3f}")
print()
print('...but strip out the market exposure the long book rents while it is in:')
print(f"   its edge over SPY is {R['cln_long_alpha']:+.2f}%/yr "
      f"(t = {R['cln_long_alpha_t']:+.2f}) -- inside the noise")
print(f"the book is live only {R['cln_invested']:.1f}% of days, "
      f"averaging {R['cln_avg_names']:.1f} names")
```

Neither direction works. On the full window the long book is indistinguishable from cash while carrying a 40% drawdown in a handful of illiquid funds. On the clean window it *looks* better — Sharpe +0.448 — but that is the stock market, not the discount: it is long ordinary equity an eighth of the time, and once you charge it for that borrowed exposure the edge is +2.98% a year with a *t* of +1.79. It still loses to simply owning SPY. The short book just loses money, at every borrow rate we tried, on both windows.

## 6. Live check — the machinery does work (offline **synthetic** data)

Before believing a null result, prove the detector can see something. We plant a rights effect into a synthetic panel — a drop on announcement, a slide through the subscription period, a bounce afterwards — and check the same code finds it. **These are synthetic numbers, not the real tape.**

```python
import os, sys
sys.path.insert(0, os.path.abspath('..'))
sys.path.insert(0, os.path.abspath(os.path.join('..','..','..')))
from rights_offering import data, strategy as st
import numpy as np
px, ev, truth = data.synthetic_panel(signal_strength=1.0, seed=929)
d = st.synthetic_detect(px, ev)
print(f"SYNTHETIC planted world ({d['n']} deals):")
for w in ('announce', 'subscription', 'post_expiry'):
    print(f"   {w:13s}{d[w+'_mean_pct']:+7.2f}%  (t={d[w+'_t']:+6.2f})")
nulls = {w: [] for w in ('announce', 'subscription', 'post_expiry')}
for s in range(8):
    p0, e0, _ = data.synthetic_panel(signal_strength=0.0, seed=929 + s)
    d0 = st.synthetic_detect(p0, e0)
    for w in nulls:
        nulls[w].append(d0[w + '_mean_pct'])
print('SYNTHETIC null world (effect switched off), averaged over 8 seeds:')
for w, vals in nulls.items():
    print(f"   {w:13s}{np.mean(vals):+7.2f}%  (spread {np.std(vals, ddof=1):.2f}%)")
```

The detector fires hard on a planted effect and, averaged across seeds, sits on zero when there is none. (Averaging matters: any single null seed can throw a two-sigma window — which is precisely the trap the real tape sets in section 4.) So the flat real-tape result is a fact about rights offerings, not a broken harness.

## Verdict

- **Signal — None.** No announcement reaction (+0.27%, *t* = +0.34), no post-expiry compensation (+0.21%, *t* = +0.25), no gradient in the depth of the discount (permutation *p* = 0.75). The single significant number lives in the window contaminated by mechanical dilution, misses the fair placebo (*p* = 0.09) and does not survive a jackknife, an era cut, a timetable sweep or a fortnight of anchor jitter.
- **Tradability — Mirage.** Long excess Sharpe +0.098 on the full window, +0.448 on the clean one — but the clean version's edge over the market is +2.98%/yr (*t* +1.79), and both lose to SPY's +0.575. The short is negative at every borrow rate.
- **Three honest caveats, and they all cut the same way.** The event list is hand-compiled and **month-precision** — which means the book is sometimes long *before* the news is public, a look-ahead that can only help it. Five further deals (CUBA, NHF, ENZN, SPLP, TUEM) have no retrievable tape at all, so the sample is **survivor-biased towards deals that ended well**. And the dilution artefact sits inside the only negative window. Every bias we know of pushes towards finding an effect, and there is still none.
