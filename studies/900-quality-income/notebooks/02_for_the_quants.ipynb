## Verdict

- **Signal — Weak.** Quality edges yield by +0.9 pp/yr and a Sharpe gap of **+0.047**, with a genuinely shallower max drawdown (**-22.4%** vs **-27.5%**) from dodging the 2020 yield-sleeve crater — but the risk-adjusted edge is **not** significant (NW *t* = **+0.57**, bootstrap CI **[-0.19, +0.24]** straddles zero) and no era is significant. A real trap-avoidance *profile*, not a certified premium. The synthetic control recovers a *planted* edge cleanly (*t* = 3.03, fires on 1/20 nulls), so a real premium *would* have shown — this one didn't clear the bar. Short single-regime tape.
- **Tradability — Fragile.** The tilt is trivially buyable (cheap, liquid, long-only, monthly rebalance costs < 1 bp/yr — not a Mirage) — but there is no significant premium to bank; you buy a shallower-drawdown risk profile, not an edge. And both dividend sleeves trailed plain SPY by ~3-4 pp/yr.
