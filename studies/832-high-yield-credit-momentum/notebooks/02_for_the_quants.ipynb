## Verdict

- **Signal — None.** The credit trend does not time equity risk-on/off: the risk-on−risk-off equity-excess difference is wrong-signed and negligible (**-1.62 bps/day**, NW *t* = **-0.37** at 6 months; -2.55 bps, *t* = -0.61 at 3 months), sits -0.4σ inside a label-shift placebo (p = 0.65), and flips sign across eras (+4.8 → -12.3 bps). The 20-seed synthetic control recovers a *planted* effect (*t* = +2.99, fires on 1/20 nulls), so the flat real-tape result is a true absence.
- **Tradability — Mirage.** The costed SPY↔IEF timer never robustly beats buy-and-hold (net Sharpe 0.627 vs 0.619 at 1 bp, active NW *t* = -1.26; it forfeits 3.5%/yr of CAGR) and falls below B&H by 5 bps. Its one merit — a -41% vs -55% drawdown — is insurance paid for in forgone return, not an edge.
