The deep companion to the [notebook for the curious](01_for_the_curious.ipynb). We test the Halloween seasonal on 98 years and show why even a durable seasonal doesn't justify the rule.

> ⚠️ **Not investment advice.** S&P 500 (^GSPC) monthly TR, 1928–2026 (Yahoo). Sources in [`docs/references.md`](../docs/references.md).

```python
import sys, os
sys.path.insert(0, os.path.abspath("../../.."))  # repo root (quantlab/)
sys.path.insert(0, os.path.abspath(".."))        # study package (summer_lull/)
%matplotlib inline
import matplotlib.pyplot as plt
plt.rcParams["figure.figsize"] = (10, 5.5); plt.rcParams["axes.grid"] = True
import numpy as np, pandas as pd
from summer_lull import data, strategy as st
ret = data.fetch_index()                       # cache-first
sp = st.seasonal_split(ret)
sim, bh = st.sell_in_may(ret), st.buy_hold(ret)
```

## Verdict, up front

| Axis | Stamp | Why |
|---|---|---|
| Signal | **Weak** | winter +5.3% vs summer +2.7%/yr, Welch t 1.3 |
| Tradability | **Mirage** | sell-in-May Sharpe 0.43 ≈ buy-and-hold 0.42, less wealth |
| Good advice? | **Not supported** | summer is positive; skipping it can't help |

> 💡 *In plain words:* a real seasonal, a self-defeating rule.

## 1 · The claim, steelmanned

- **H₁:** winter returns exceed summer significantly.
- **H₂:** the sell-in-May rule beats buy-and-hold.
- **H₃:** the seasonal is stable over time.

## 2 · So what? — what rides on each

H₁ can hold while H₂ fails — that's the whole study. A weaker-but-positive summer means the rule forgoes return for no Sharpe gain.

## 3 · How we'd know — the protocol

Winter vs summer means + Welch t → sell-in-May vs buy-and-hold → pre/post-2000 persistence → confirm summer is positive.

## 4 · The teardown

### 4.1 The seasonal split

```python
print({k:(round(v,2) if isinstance(v,float) else v) for k,v in sp.items()})
print(f"winter {sp['winter_ann']:+.1%}/yr vs summer {sp['summer_ann']:+.1%}/yr — Welch t {sp['welch_t']:.2f}")
```

> 💡 *In plain words:* winter wins, but t 1.3 is soft and summer is **positive**. H₁ weakly holds.

### 4.2 The rule vs buy-and-hold

```python
display(pd.DataFrame({'sell-in-May':st.summary(sim),'buy & hold':st.summary(bh)}).T[['cagr','sharpe','vol_ann','max_drawdown']].round(3))
```

> 💡 *In plain words:* equal Sharpe, a third less wealth — the rule only cut exposure. **H₂ rejected.**

### 4.3 A century of persistence

```python
import numpy as np
for lab,lo,hi in [('1928-1999',1900,1999),('2000-on',2000,2100)]:
    sl=ret[(ret.index.year>=lo)&(ret.index.year<=hi)]; d=st.seasonal_split(sl)
    print(f'{lab}: winter {d["winter_bp"]:.0f} bp/mo vs summer {d["summer_bp"]:.0f} bp/mo')
```

> 💡 *In plain words:* the gap survives post-publication (Jacobsen-Zhang 2018) — rare for a calendar effect. **H₃ holds** — and *still* doesn't rescue the trade.

## 5 · The verdict

H₁ weak, H₃ holds, H₂ rejected → Signal `WEAK`, Tradability `MIRAGE`, the advice `NOT SUPPORTED`.

## 6 · Could you trade it?

No edge over buy-and-hold. The durable seasonal is a fact about *when* returns accrue, not a reason to be in cash through a positive season.
