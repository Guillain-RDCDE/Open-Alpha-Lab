## Verdict

- **Signal — Weak.** A **real, robust drawdown & vol cut** (max DD -12.5% vs -30.8%, positive in both eras) — but the *Sharpe advantage* (+0.13) has a bootstrap CI that straddles zero ([-0.24, +0.52], P>0 = 0.74), the return difference is negative & insignificant (-0.87 bps/day, NW *t* = -1.18), and it is front-loaded on 2008 (era adv +0.20 → +0.07). Real risk reduction, no robust risk-adjusted outperformance. *Short history flatters the trend rule.*
- **Tradability — Fragile.** Trading costs are trivial (+0.08 adv at 5 bps), so a tax-deferred account gets a genuinely calmer ride — but a **15% short-term tax flips the advantage negative** (-0.09), 25% takes it to -0.28, and the edge decays across eras. Real-but-thin & tax-eaten — Fragile, not Investable, not a Mirage (the drawdown protection does survive costs).
