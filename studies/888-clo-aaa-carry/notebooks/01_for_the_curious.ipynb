# Study 888 — CLO AAA Carry 🔒

**Does the senior slice of a CLO pay a *real* pickup over cash and same-rated corporates?**

A **AAA-rated CLO tranche** sits at the top of a collateralised-loan-obligation waterfall:
first to be paid, last to take losses, protected by a thick cushion of subordinate tranches.
It pays a spread over cash and over same-rated IG corporate bonds — the story goes — as
compensation for *structural complexity* (a securitisation few desks underwrite), not for
credit risk (realized senior-tranche defaults are ~nil). Since 2020-10 it trades in a liquid,
floating-rate ETF (**JAAA**; **ICLO** from 2022-12), so it carries almost no *duration* either.

We test that carry **excess-of-cash** (minus BIL) against IG corporates (**LQD**), Treasuries
(**IEF**), and — the sharp control — the *un-tranched* leveraged-loan collateral itself
(**BKLN**). 2020-10-20 → 2026-06-30.

*Numbers below are the frozen headline (`docs/results.md`); the live cells run the fast
synthetic control. **Short history:** JAAA ~5.7y, ICLO ~3.5y — ONE rate cycle, NO CLO
credit-stress event in-sample, so realized Sharpes are an upper bound (named on the Signal
axis). This is distinct from **[614-clo-equity-yield](../../614-clo-equity-yield/)**, the
risky first-loss BOTTOM of the same stack.*

## 3. Is the detector honest? A live synthetic control

We plant a steady +1.2%/yr excess-of-cash carry in a seeded toy world (plus a high-vol 'duration' decoy that earns *nothing* extra) and check the detector recovers it — and stays silent on the null (`carry=0`). No network.

## 4. The honest verdict

**Signal — Real (but thin & regime-bound).** The AAA-CLO carry *is* a genuine risk-adjusted pickup: JAAA earns **+1.38%/yr** over cash on just **1.63%** vol (**Sharpe +0.84**, HAC *t* +2.33, bootstrap CI clear of zero), dominating IG corporates (Sharpe -0.11), Treasuries (-0.33) *and* the un-tranched loan collateral (+0.24) — so the seniority earns real keep. ICLO independently confirms (+0.88). **But** it's a regime bet: in the ZIRP era the excess was **+0.00%/yr** (nothing to pocket when cash is 0); all of it is the +1.84%/yr high-rate era. And the ~5.7y sample has **no CLO stress event** — the one AAA-CLO mark-down (Mar-2020) predates JAAA.

**Tradability — Fragile.** It survives costs trivially (buy-and-hold, the ~0.20% ER already in the NAV, tight spreads → net **+1.35%/yr**) and capacity is ample (JAAA is a >$20bn fund). What makes it *Fragile* not *Investable*: the whole ~1.4%/yr IS an insurance premium for a tail (illiquidity/credit) the calm sample never charged — a single crisis mark-down (AAA CLOs fell ~5-10% in Mar-2020) would erase years of carry, so the realized Sharpe flatters the true edge.
