The deep companion to the [notebook for the curious](01_for_the_curious.ipynb). We build gross-profitability quality on SEC data and show it points right but can't be confirmed here.

> ⚠️ **Not investment advice.** EDGAR GrossProfit/Assets + Yahoo annual returns, current S&P 500 members, ~2007–2025 (survivorship-biased, large-cap, short). Sources in [`docs/references.md`](../docs/references.md).

```python
import sys, os
sys.path.insert(0, os.path.abspath("../../.."))  # repo root (quantlab/)
sys.path.insert(0, os.path.abspath(".."))        # study package (blue_chip/)
%matplotlib inline
import matplotlib.pyplot as plt
plt.rcParams["figure.figsize"] = (10, 5.5); plt.rcParams["axes.grid"] = True
import numpy as np, pandas as pd
from blue_chip import data, strategy as st
sig, fwd = data.fetch_panel()                  # cache-first (shared EDGAR pull)
h = st.quantile_hedge(sig, fwd, q=0.2, long_high=True)
```

## Verdict, up front

| Axis | Stamp | Why |
|---|---|---|
| Signal | **Weak** | hedge +3.3%/yr, Sharpe 0.20, t ≈ 0.8 |
| Tradability | **Fragile** | ~18 XBRL years, survivorship, large-cap |
| Quality real? | **Supported** | right sign + Novy-Marx (2013) |

> 💡 *In plain words:* a real factor we can only glimpse on free data.

## 1 · The claim, steelmanned

- **H₁:** high-GP firms out-earn low-GP firms (positive hedge).
- **H₂:** it's significant.
- **H₃:** it's a known, durable factor.

## 2 · So what? — what rides on each

If H₁/H₂ hold on free data, quality is a confirmed, tradable tilt. If only H₁/H₃ hold, the factor is real but our window is too short to prove it.

## 3 · How we'd know — the protocol

GP/Assets → annual long-high/short-low → the hedge's mean, Sharpe, t-stat and hit rate → the two legs vs the universe.

## 4 · The teardown

### 4.1 The hedge

```python
display(h.round(3))
s=st.summary(h['hedge']); print(f"hedge: mean {s['mean']:+.2%}/yr  Sharpe {s['sharpe']:.2f}  t≈{s['tstat']:.1f}  hit {s['hit_rate']:.0%}  n={s['n']}")
```

> 💡 *In plain words:* positive (+3.3%/yr, hit 61%) — **H₁ holds** — but t ≈ 0.8, **H₂ rejected** on this sample.

### 4.2 The two legs vs the universe

```python
print('high-GP leg : %+.2f%%' % (st.summary(h['high'])['mean']*100))
print('low-GP leg  : %+.2f%%' % (st.summary(h['low'])['mean']*100))
print('universe    : %+.2f%%' % (st.summary(st.market_annual(fwd).reindex(h.index))['mean']*100))
```

> 💡 *In plain words:* high-GP beats low-GP by ~3 pts/yr, both near the universe — a mild tilt, the right way.

### 4.3 Why ~18 years under-powers a real effect

SEC machine-readable (XBRL) filings start ~2007, so we miss the 1963–2007 decades where Novy-Marx documents the strong quality premium. With ~18 annual observations, even a genuine effect carries a t near 0.8. Add survivorship (current members) and a large-cap-only universe (quality's short leg is often smaller, distressed names), and the *test* is under-powered — not the factor. **H₃ supported by sign + literature.**

## 5 · The verdict

H₁ holds, H₂ rejected (sample), H₃ supported → Signal `WEAK`, Tradability `FRAGILE`, quality premium `SUPPORTED`.

## 6 · Could you trade it?

Yes, via low-cost quality ETFs — it's the most defensible factor tilt. The honest caveat is confirmation power on free data, not the premium's existence.
