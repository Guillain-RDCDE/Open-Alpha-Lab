## Verdict, up front

> 💡 *In plain words:* a genuine factor, measured in the wrong universe.

## 1 · The claim, steelmanned

- **H₁:** high net issuance predicts low forward returns.
- **H₂:** the long-buyback / short-issuer hedge is positive and significant.
- **H₃:** it's tradable on liquid large caps (not just micro-caps).

## 2 · So what? — what rides on each

H₁/H₂ would make issuance a free, fundamentals-based signal straight from 10-Ks. H₃ decides whether it survives where you can actually trade. A small-cap-only effect is real science but useless to a large-cap book.

## 3 · How we'd know — the protocol

EDGAR diluted shares → net issuance (YoY change) → annual quintile sort → long bottom-20% (buyers), short top-20% (issuers) → next-year hedge, Sharpe, t-stat, hit rate over the XBRL window.

## 4 · The teardown

### 4.1 The long-short, year by year

```python
disp=h.copy(); disp.columns=['issuer leg','buyer leg','hedge (buyer-issuer)']
display((disp*100).round(1))
s=st.summary(h['hedge']); print({k:round(v,3) for k,v in s.items()})
```

> 💡 *In plain words:* the hedge mean is **negative** (−2.8%/yr) and wins only ~28% of years. **H₁ and H₂ are rejected on this universe** — the sign is backwards.

### 4.2 Why the sign inverts — H₃ fails

Three forces flip it on large caps:

1. **Issuers were the winners.** 2010–2025's heavy large-cap issuers were high-growth tech diluting via stock-based comp and M&A — the market leaders.
2. **The anomaly lives in small caps.** Pontiff-Woodgate's effect concentrates in small, illiquid names; Hou-Xue-Zhang (2020) show financing anomalies fade with value-weighting and NYSE breakpoints.
3. **Survivorship.** Current members exclude the dilutive blow-ups that delisted — which would have helped the short — yet even this biased test can't find the effect. **H₃ rejected.**

## 5 · The verdict

H₁/H₂/H₃ all rejected on tradable large caps → Signal `WEAK` (inverted), Tradability `MIRAGE`, anomaly-on-large-caps `NOT SUPPORTED`. The fingerprinted run is in [`docs/results.md`](../docs/results.md).

## 6 · Could you trade it?

Not on this universe. To harvest the real premium you'd need a small-cap, value-weighted, NYSE-breakpoint book with the delisted names included — outside what a free EDGAR pull on current S&P members can offer, and outside what most large books can size.
