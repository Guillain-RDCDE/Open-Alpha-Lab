## Verdict, up front

## 1 · The claim, steelmanned

- **H₁:** longer Treasuries earn a positive excess over cash (the term premium).
- **H₂:** extending duration improves the Sharpe.
- **H₃:** the duration risk is mild.

## 2 · So what? — what rides on each

H₁ is the premium's existence; H₂/H₃ decide whether reaching for it is wise. A real-but-poorly-paid premium argues for the short end.

## 3 · How we'd know — the protocol

The three-point duration ladder (CAGR/Sharpe/vol/DD) → the isolated term premium (IEF−BIL) with its t-stat and drawdown.

## 4 · The teardown

### 4.1 The duration ladder — return up, Sharpe down

```python
import pandas as pd
display(pd.DataFrame({'IEF (7-10y)':st.leg_summary(al,'IEF'),'SHY (1-3y)':st.leg_summary(al,'SHY'),'BIL (cash)':st.leg_summary(al,'BIL')}).T[['cagr','sharpe','vol_ann','max_drawdown']].round(3))
```

> 💡 *In plain words:* CAGR rises with duration (**H₁ holds**) but Sharpe falls 1.82 → 0.52 (**H₂ rejected**) and the 7–10y drawdown is −23% (**H₃ rejected**).

### 4.2 The isolated term premium

```python
print({k:round(v,3) for k,v in st.excess_stats(tp).items()})
ld=tp.loc['2022']; print('2022 term-premium return:', f'{((1+ld).prod()-1):+.1%}')
```

> 💡 *In plain words:* +2.2%/yr at Sharpe 0.32 (t 1.4), with a brutal 2022 — a thin, lumpy carry.

### 4.3 Why duration disappointed

(1) **Convex, lumpy risk** — quiet roll-down then a −17% year; (2) **compressed premia** — QE/falling rates pushed term premia toward zero (Cochrane-Piazzesi 2005); (3) **the short end is efficient** — bills/1–3y captured most of the safe yield at a fraction of the risk.

## 5 · The verdict

H₁ holds, H₂/H₃ rejected → Signal `REAL`, Tradability `FRAGILE`, the duration trade `NOT SUPPORTED`.

## 6 · Could you trade it?

For most cautious investors the short end dominated risk-adjusted. Extend duration with a *rate view*, not for the carry — the premium didn't pay for its risk.
